```python
# 6. Count the digits in a number
n = abs(int(input("Enter an integer: ")))
count = 1 if n == 0 else 0
while n > 0:
    count += 1
    n //= 10
print("Number of digits:", count)
```

```python
# 7. Reverse a number using a loop
n = int(input("Enter an integer: "))
sign = -1 if n < 0 else 1
n = abs(n)
reverse = 0
while n > 0:
    reverse = reverse * 10 + n % 10
    n //= 10
print("Reversed number:", sign * reverse)
```

```python
# 8. Check whether a number is prime
n = int(input("Enter an integer: "))
is_prime = n >= 2
for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
        is_prime = False
        break
print("Prime" if is_prime else "Not prime")
```

```python
# 9. Function to check whether a number is a palindrome
def is_palindrome(n):
    text = str(n)
    return text == text[::-1]

number = int(input("Enter a number: "))
print("Palindrome" if is_palindrome(number) else "Not palindrome")
```

```python
# 10. Function to calculate the sum of digits
def sum_of_digits(n):
    n = abs(n)
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total
```

```python
# 27. Rectangle class to calculate area and perimeter
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rectangle = Rectangle(10, 5)
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
```

```python
# 28. Employee class with salary-increase method
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, amount):
        self.salary += amount

employee = Employee("Neha", 30000)
employee.increase_salary(5000)
print(employee.name, employee.salary)
```

```python
# 29. ShoppingCart class using a dictionary (item: price)
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def total_bill(self):
        return sum(self.items.values())

cart = ShoppingCart()
cart.add_item("Book", 250)
cart.add_item("Pen", 20)
print("Items:", cart.items)
print("Total bill:", cart.total_bill())
```

```python
# 30. Library class to add and remove books
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)
        else:
            print("Book not found")

library = Library()
library.add_book("Python Basics")
library.add_book("Machine Learning")
library.remove_book("Python Basics")
print("Books:", library.books)
```
