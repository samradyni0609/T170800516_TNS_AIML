### Exercise 2: Seaborn line plot
Write a Python program using Seaborn to create a line plot.

- Days: `1, 2, 3, 4, 5`
- Temperature: `30, 32, 31, 35, 36`


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [200, 300, 250, 400]

plt.plot(months, sales, marker="o")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 36]

sns.lineplot(x=days, y=temperature, marker="o")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Daily Temperature")
plt.grid(True)
plt.show()
```

```python
height = [150, 160, 170, 180]
weight = [50, 60, 65, 80]

sns.scatterplot(x=height, y=weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```
