```python
#Day 1 – Python Basics and Object-Oriented Programming (OOP)
#1. Variables and Data Types
name = "Samradyni"
age = 12
percentage = 78.90
is_student = True

print(name)
print(type(age))
print(type(percentage))
```

```python
#2. Operators

a = 5
b = 7

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a == b)
```

```python
#3. Conditional Statements

marks = 65

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

```python
#4. Loops

for i in range(5):
    print(i)
```

```python
#5. String Operations

text = "Python"

print(text.upper())
print(text.lower())
print(text[0:3])
print(text[::-1])
```

```python
#6. Functions

def add(x, y):
    return x + y

result = add(5, 6)
print(result)
```

```python
# Simple Prediction

x = float(input("Enter value of x: "))
y = 2 * x + 3

print("Predicted y:", y)
```

```python
#8. Threshold-Based Classification

temp = float(input("Enter temperature: "))

if temp > 30:
    print("Hot")
else:
    print("Normal")
```

```python
#9. Object-Oriented Programming (OOP)

class Dog:
    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.age)
print(dog1.species)
```
