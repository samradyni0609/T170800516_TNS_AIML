## 2. Questions or Exercises Mentioned in the Notebook


Write a Python program using NumPy for the following marks: `65, 78, 90, 55, 88`.

Tasks:
1. Create a NumPy array using these marks.
2. Find the average marks.
3. Find the highest marks.
4. Find the lowest marks.



```python
import numpy as np

marks = np.array([65, 78, 90, 55, 88])

print("Marks:", marks)
print("Average marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
```

```python
import numpy as np

arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)

print("Array:", arr)
print("3 x 4 matrix:\n", matrix)
print("Transpose:\n", matrix.T)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Sara", "John"],
    "Age": [20, 21, 22],
    "Marks": [85, 90, 88]
})

print(df)
print(df["Name"])
print(df[["Name", "Marks"]])
print(df.iloc[0])
```
